Step 7: Separate Input Features (X) and Target Variable (y)

```python
# Step 7.1: Separate input features and target variable

# X contains all input features
X = df.drop("Exam_Score", axis=1)

# y contains the value we want to predict
y = df["Exam_Score"]

# Display the shapes
print("Input Features (X) Shape:", X.shape)
print("Target Variable (y) Shape:", y.shape)

# Display the first five rows of X
print("\nFirst Five Rows of Input Features:")
display(X.head())

# Display the first five target values
print("\nFirst Five Target Values:")
display(y.head())
```

```python
# Step 7.2: Identify numerical and categorical features

numerical_features = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("Numerical Features:")
print(numerical_features)

print("\nCategorical Features:")
print(categorical_features)

print("\nTotal Numerical Features:", len(numerical_features))
print("Total Categorical Features:", len(categorical_features))
```

Step 8: Training and Testing Data

```python
# Step 8.1: Import train_test_split

from sklearn.model_selection import train_test_split

print("train_test_split imported successfully!")
```

```python
# Step 8.2: Split the dataset into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Dataset split completed successfully!")

print("\nTraining Features:", X_train.shape)
print("Testing Features:", X_test.shape)

print("\nTraining Target:", y_train.shape)
print("Testing Target:", y_test.shape)
```

Step 9: Data Preprocessing

```python
# Step 9.1: Import preprocessing tools

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

print("Preprocessing libraries imported successfully!")
```

```python
# Step 9.2: Create preprocessing pipelines

# Handle numerical features
numerical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median"))
])

# Handle categorical features
categorical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

print("Numerical and categorical pipelines created successfully!")
```

```python
# Step 9.3: Combine numerical and categorical preprocessing

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numerical_pipeline, numerical_features),
        ("cat", categorical_pipeline, categorical_features)
    ]
)

print("Complete preprocessing system created successfully!")
```

```python
# Step 9.4: Fit preprocessing on training data and transform both sets

X_train_processed = preprocessor.fit_transform(X_train)

X_test_processed = preprocessor.transform(X_test)

print("Data preprocessing completed successfully!")

print("\nOriginal Training Shape:", X_train.shape)
print("Processed Training Shape:", X_train_processed.shape)

print("\nOriginal Testing Shape:", X_test.shape)
print("Processed Testing Shape:", X_test_processed.shape)
```

Step 10: Train the Linear Regression Model

```python
# Step 10.1: Import Linear Regression

from sklearn.linear_model import LinearRegression

print("Linear Regression imported successfully!")
```

```python
# Step 10.2: Create the Linear Regression model

model = LinearRegression()

print("Linear Regression model created successfully!")
```

```python
# Step 10.3: Train the Linear Regression model

model.fit(X_train_processed, y_train)

print("Model training completed successfully!")
```

```python
# Step 10.4: Examine the trained model

print("Model Intercept:")
print(model.intercept_)

print("\nNumber of Model Coefficients:")
print(len(model.coef_))

print("\nModel Training Completed!")
```

Step 11: Evaluate Our Machine Learning Model

```python
# Step 11.1: Make predictions on the testing dataset

y_pred = model.predict(X_test_processed)

print("Predictions generated successfully!")

print("\nFirst 10 Actual Exam Scores:")
print(y_test.head(10).values)

print("\nFirst 10 Predicted Exam Scores:")
print(y_pred[:10])
```

```python
# Step 11.2: Import evaluation metrics

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# Calculate MAE
mae = mean_absolute_error(y_test, y_pred)

# Calculate MSE
mse = mean_squared_error(y_test, y_pred)

# Calculate RMSE
rmse = np.sqrt(mse)

# Calculate R-squared
r2 = r2_score(y_test, y_pred)

print("Model Evaluation Results")
print("-" * 35)

print(f"Mean Absolute Error (MAE): {mae:.4f}")
print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.4f}")
print(f"R-squared Score (R²): {r2:.4f}")
```

```python
# Step 11.3: Compare actual and predicted exam scores

comparison = pd.DataFrame({
    "Actual Score": y_test.values,
    "Predicted Score": y_pred
})

comparison["Predicted Score"] = comparison["Predicted Score"].round(2)

comparison["Prediction Error"] = (
    comparison["Actual Score"] - comparison["Predicted Score"]
).round(2)

display(comparison.head(10))
```

Step 12: Visualize Actual vs. Predicted Exam Scores

```python
# Step 12: Visualize Actual vs. Predicted Exam Scores

plt.figure(figsize=(8, 6))

# Compare actual scores with predicted scores
plt.scatter(y_test, y_pred, alpha=0.6)

# Draw a reference line for perfect predictions
min_score = min(y_test.min(), y_pred.min())
max_score = max(y_test.max(), y_pred.max())

plt.plot(
    [min_score, max_score],
    [min_score, max_score],
    linestyle="--"
)

# Add graph labels and title
plt.xlabel("Actual Exam Scores")
plt.ylabel("Predicted Exam Scores")
plt.title("Actual vs. Predicted Student Exam Scores")

# Display the graph
plt.tight_layout()
plt.show()
```

Step 13: Analyze Feature Importance

```python
# Step 13: Analyze Feature Importance

# Get the names of the features after preprocessing
feature_names = preprocessor.get_feature_names_out()

# Get the coefficients learned by Linear Regression
coefficients = model.coef_

# Create a DataFrame to organize the results
feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Coefficient": coefficients
})

# Calculate the absolute coefficient values
feature_importance["Absolute Importance"] = (
    feature_importance["Coefficient"].abs()
)

# Sort features from highest to lowest absolute coefficient
feature_importance = feature_importance.sort_values(
    by="Absolute Importance",
    ascending=False
)

# Display the 15 features with the largest absolute coefficients
print("Top 15 Features by Absolute Coefficient")
display(feature_importance.head(15))
```

```python
# Display the top 15 features in a bar chart

top_features = feature_importance.head(15).sort_values(
    by="Absolute Importance"
)

plt.figure(figsize=(10, 7))

plt.barh(
    top_features["Feature"],
    top_features["Absolute Importance"]
)

plt.xlabel("Absolute Coefficient Value")
plt.ylabel("Features")
plt.title("Top 15 Features by Absolute Coefficient")
plt.tight_layout()
plt.show()
```

Goal of Step 14

```python
# Step 14: Train a Random Forest Model

from sklearn.ensemble import RandomForestRegressor

# Create the Random Forest model
random_forest_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

# Train the model using the preprocessed training data
random_forest_model.fit(X_train_processed, y_train)

print("Random Forest model trained successfully!")
```

```python
# Generate predictions from both models

linear_predictions = model.predict(X_test_processed)

random_forest_predictions = random_forest_model.predict(
    X_test_processed
)

# Evaluate both models
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

linear_mae = mean_absolute_error(y_test, linear_predictions)
linear_rmse = np.sqrt(mean_squared_error(y_test, linear_predictions))
linear_r2 = r2_score(y_test, linear_predictions)

rf_mae = mean_absolute_error(y_test, random_forest_predictions)
rf_rmse = np.sqrt(
    mean_squared_error(y_test, random_forest_predictions)
)
rf_r2 = r2_score(y_test, random_forest_predictions)

# Create a comparison table
model_comparison = pd.DataFrame({
    "Model": ["Linear Regression", "Random Forest"],
    "MAE": [linear_mae, rf_mae],
    "RMSE": [linear_rmse, rf_rmse],
    "R2 Score": [linear_r2, rf_r2]
})

print("Model Performance Comparison")
display(model_comparison.round(4))
```

Step 15: Select the Best Model

```python
# Step 15: Select the Best Model

# Compare the MAE values of both models
if linear_mae <= rf_mae:
    best_model = model
    best_model_name = "Linear Regression"
else:
    best_model = random_forest_model
    best_model_name = "Random Forest"

# Display the selected model
print("Best Model Selected:", best_model_name)

# Display its evaluation results
if best_model_name == "Linear Regression":
    print(f"MAE: {linear_mae:.4f}")
    print(f"RMSE: {linear_rmse:.4f}")
    print(f"R2 Score: {linear_r2:.4f}")
else:
    print(f"MAE: {rf_mae:.4f}")
    print(f"RMSE: {rf_rmse:.4f}")
    print(f"R2 Score: {rf_r2:.4f}")
```

Step 16: Save the Trained Model

```python
# Step 16: Save the Trained Model

import joblib
import os

# Create a folder to store the trained model
os.makedirs("models", exist_ok=True)

# Save the selected model
joblib.dump(best_model, "models/student_performance_model.pkl")

# Save the preprocessing pipeline
joblib.dump(preprocessor, "models/preprocessor.pkl")

print("Model saved successfully!")
print("Preprocessor saved successfully!")
```
